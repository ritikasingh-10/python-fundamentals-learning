#### WAP to check if a list contains a palindrome of elements.

```python
list1 = [1,2,1,]

copy_list1 = list1.copy()
copy_list1.reverse()

if(copy_list1 == list1):
    print("its palidrome")
else:
    print("not palidrome")
```

#### Level 2: Indexing, Slicing & Methods

```python
'''
6. List Slicing
Given:
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
Print:
First 3 elements
Last 3 elements
Elements from index 2 to 5
Every second element
List in reverse
'''

numbers = [10, 20, 30, 40, 50, 60, 70, 80]
print(numbers[0:4]," <-First 3 elements" )
print(numbers[-4:-1]," <-Last 3 elements")
print(numbers[2:5]," <- Elements from index 2 to 5")
print(numbers[::2]," <- Every second element")
print(numbers[::-1],"<- reverse")
print(numbers,"<- original list")
```

```python
#10. Sort a List
'''
Marks in ascending order
Marks in descending order
Highest mark
Lowest mark
'''

marks = [78, 45, 92, 61, 85, 39, 70]

#Marks in ascending order
marks.sort()
print(marks)

#Marks in descending order
marks.sort(reverse = True)
print(marks)

#Highest mark
print(max(marks))

#Lowest mark
print(min(marks))
```
