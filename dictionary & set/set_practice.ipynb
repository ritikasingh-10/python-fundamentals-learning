```python
## 7. Find Common Interests -- using intersection
person1 = {"Python", "SQL", "Excel", "Reading"}
person2 = {"Python", "Java", "SQL", "Gaming"}

common = person1.intersection(person2)
print(common)
```

```python
## 8. Find Unique Skills
skills1 = {"Python", "SQL", "Excel"}
skills2 = {"Python", "Java", "C++"}

unique = skills1.difference(skills2)
print(unique)
```

```python
## 9. Find All Skills -- Using the same two sets, find all unique skills possessed by both people.
set_1 = {"Python", "SQL", "Excel", "Java", "C++"}
set_2 = {"Python", "SQL", "Excel", "C++"}

unique = set_1.union(set_2)
print(unique)
```
