#### 1. Create and Access a Dictionary
Create a dictionary containing information about yourself:
Name
Age
Course
City

Print:
Complete dictionary
Name
Course
City
Concept: dictionary creation, keys, values, accessing values.

```python
mydic = {"name" :"ritika singh",
         "age" : 20,
         "course" : "DSAI",
         "City":"mumbai"
        }
print(mydic)
print(mydic["name"])
print(mydic["course"])
print(mydic["City"])
```

#### 2. Add a New Key-Value Pair
Add:
"college": "ABC College"

```python
student = {
    "name": "Ritu",
    "age": 21,
    "course": "Data Science"
}
print(student)

student["college"] = "ABC college"
print(student)
```

#### 3. Update a Value
Change the course to "Data Science".
Concept: updating dictionary values.

```python
student = {
    "name": "Ritu",
    "age": 21,
    "course": "Python"
}

print("original dict :",student)
student["course"] = "Data science"
print("updated value:",student)
```

#### 4. Delete a Key
Remove the "age" key.
Print the updated dictionary.
Concept: del or pop().

```python
employee = {
    "name": "Aman",
    "age": 25,
    "department": "IT",
    "salary": 50000
}

del employee["age"]
print(employee)
```

#### Level 3: Dictionary + Conditions

11. Check Student Result
Check whether the student passed all subjects.

Rule:

Marks must be at least 40 in every subject.

```python
student = {
    "name": "Ritu",
    "maths": 78,
    "python": 85,
    "sql": 72
}

if student["maths"] >= 40 and student["python"] >= 40 and student["sql"] >= 40:
    print("pass")
else:
    print("failed")
```

#### 12. Find the Highest Marks
Find:

Highest marks
Subject with the highest marks

```python
marks = {
    "Maths": 78,
    "Python": 92,
    "SQL": 85,
    "Statistics": 74
}

highest = max(marks["Maths"],marks["Python"],marks["SQL"],marks["Statistics"])

if highest==marks["Maths"]:
    subject = "Maths"
elif highest==marks["Python"]:
    subject = "Python"
elif highest==marks["SQL"]:
    subject = "SQL"
else :
    subject = "Statisctics"

print("Highest marks:", highest)
print("Subject:", subject)
```
