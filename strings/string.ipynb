```python
##A palindrome is a word that remains the same when reversed.
word = input("enter the word")
reverse_indexing = word[::-1]
print("reverse word:",reverse_indexing)
if word==reverse_indexing:
    print("its palindrome ")
else:
    print("its not palindrome")
```

### Create a simple password checker.

```python
password = input("Enter password: ")

if len(password) >= 8 and password != password.lower() and password != password.upper():
    print("Password is strong")
else:
    print("Password is weak")
```
