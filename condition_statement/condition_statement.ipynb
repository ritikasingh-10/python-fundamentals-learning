```python
#7. Grade Calculator--Take marks from the user and assign a grade:
'''
90–100 → A
75–89  → B
60–74  → C
40–59  → D
Below 40 → F
'''

marks = int(input("enter the marks"))
if(marks>=90):
    print("that marks come under grade A")
elif(marks>=75 and marks<=89):
    print("that marks comes under grade B")
elif(marks>=60 and marks<=74):
    print("that marks comes under grade C")
elif(marks>=40 and marks<=59):
    print("that marks comes under grade D")
else:
    print("FAIL")
```

```python
'''
12. Discount Calculator
An online store gives discounts based on purchase amount:

₹5000 or above → 20% discount
₹3000–₹4999   → 10% discount
₹1000–₹2999   → 5% discount
Below ₹1000   → No discount

Take the purchase amount and calculate the final amount after discount.
'''
amount = float(input("enter the number:"))
if(amount >= 5000):
    discount= 20
elif(amount>=3000 and amount<=4999):
    discount=10
elif(amount>=1000 and amount<=2999):
    discount=5
else:
    print("no discount")

discount_amount = amount * discount/100
final_amount = amount-discount 

print("discount:",discount,"%")
print("discount amount:",discount_amount)
print("final_amount:",final_amount)
```
